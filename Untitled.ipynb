```python
# ================================
# 1. Import des librairies
# ================================
import pandas as pd
import re
from sklearn.preprocessing import LabelEncoder


# ================================
# 2. Chargement du dataset
# ================================
file_path = "Apache_2k.log_structured11.csv"

df = pd.read_csv(file_path, sep=";")

print("=== Aperçu du dataset brut ===")
print(df.head())
print("\n=== Infos ===")
print(df.info())


# ================================
# 3. Vérification / correction des colonnes
# ================================
expected_columns = [
    "LineId",
    "Time",
    "Level",
    "Content",
    "EventId",
    "EventTemplate"
]

df.columns = expected_columns


# ================================
# 4. Nettoyage de la colonne Time
# ================================
df["Time"] = pd.to_datetime(df["Time"], errors="coerce")

print("\nValeurs Time nulles :", df["Time"].isna().sum())


# ================================
# 5. Nettoyage de la colonne Level
# ================================
df["Level"] = df["Level"].str.lower().str.strip()

print("\nRépartition des niveaux de logs :")
print(df["Level"].value_counts())


# ================================
# 6. Nettoyage du texte (Content & EventTemplate)
# ================================
def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.lower()
    text = re.sub(r"\d+\.\d+\.\d+\.\d+", "ip", text)   # adresses IP
    text = re.sub(r"\d+", "num", text)                # nombres
    text = re.sub(r"[^\w\s]", "", text)               # ponctuation
    text = re.sub(r"\s+", " ", text).strip()          # espaces multiples
    return text

df["Content_clean"] = df["Content"].apply(clean_text)
df["EventTemplate_clean"] = df["EventTemplate"].apply(clean_text)

print("\n=== Texte nettoyé (exemple) ===")
print(df[["Content_clean", "EventTemplate_clean"]].head())


# ================================
# 7. Suppression des lignes incomplètes
# ================================
df.dropna(subset=["Time", "Level", "Content"], inplace=True)

print("\nNombre de lignes après nettoyage :", len(df))


# ================================
# 8. Encodage de Level (pour ML)
# ================================
label_encoder = LabelEncoder()
df["Level_encoded"] = label_encoder.fit_transform(df["Level"])

print("\nClasses encodées :", label_encoder.classes_)


# ================================
# 9. Sauvegarde du dataset nettoyé
# ================================
output_file = "Apache_logs_cleaned_final.csv"
df.to_csv(output_file, index=False)

print(f"\n✅ Dataset nettoyé sauvegardé : {output_file}")
```
